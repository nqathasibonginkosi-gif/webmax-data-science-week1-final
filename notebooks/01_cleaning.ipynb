```python
import pandas as pd

df = pd.read_csv("leads.csv")
print("Raw rows:", len(df))
print(df.head())
print("\n--- Status BEFORE ---")
print(df['status'].value_counts(dropna=False))
print("\n--- Source BEFORE ---")
print(df['source'].value_counts(dropna=False))
```

```python
# --- CLEANING ---
print("Rows before cleaning:", len(df))

# 1. Remove spaces, make text neat
df['status'] = df['status'].str.strip().str.title()
df['source'] = df['source'].str.strip().str.title()

# 2. Fix dates - with fix for messy dates
df['lead_created_at'] = pd.to_datetime(df['lead_created_at'], errors='coerce', dayfirst=True, format='mixed')

# 3. Drop exact duplicates
df = df.drop_duplicates()
print("Rows after dropping duplicates:", len(df))

# 4. Check for missing values
print("\n--- Missing values AFTER ---")
print(df.isnull().sum())

# 5. Save the clean file
df.to_csv("leads_cleaned.csv", index=False)
print("\nSaved as leads_cleaned.csv!")

print("\n--- Status AFTER ---")
print(df['status'].value_counts(dropna=False))
print("\n--- Source AFTER ---")
print(df['source'].value_counts(dropna=False))
```
